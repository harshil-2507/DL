# Q2. **Linear Alegebra** in tensorflow

```python
import tensorflow as tf

A = tf.constant([[1, 2],  [3, 4]], dtype=tf.float32)
B = tf.constant([[5, 6],[7, 8]], dtype=tf.float32)

print("Addition:\n", tf.add(A, B))
print("Matrix Multiply:\n", tf.matmul(A, B)) #matmul

print("Transpose:\n", tf.transpose(A))

#sub functions in lialg
print("Determinant:", tf.linalg.det(A))
print("Inverse:\n", tf.linalg.inv(A))
```

# Q3. hard coded with just numpy for ***AND & OR Gates using Perceptron***

```python
import numpy as np

def perceptron(x, w, b):
    return 1 if np.dot(x, w) + b >= 0 else 0

inputs = np.array([[0,0], [0,1], [1,0], [1,1]])



# AND
w_and = np.array([1, 1])
b_and = -1.5

print("AND op")
for x in inputs:
    print(x, perceptron(x, w_and, b_and))

# OR
w_or = np.array([1, 1])
b_or = -0.5

print("\nOR op")
for x in inputs:
    print(x, perceptron(x, w_or, b_or))
```

# **Q4. XOR Problem using PyTorch Neural NW**

```python
import torch
import torch.nn as nn
import torch.optim as optim

#ip
X = torch.tensor([[0.,0.], [0.,1.], [1.,0.], [1.,1.]])
Y = torch.tensor([[0.], [1.], [1.], [0.]])

model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
    nn.Sigmoid()
)

#loass and optimizing
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.1)

#train for 2000 epochs
for epoch in range(2000):
    optimizer.zero_grad()
    output = model(X)
    loss = criterion(output, Y)
    loss.backward()
    optimizer.step()



print("XOR op:")
print(model(X).detach())
```

# **Q.5 Regressive Simple NN**

```python
import torch
import torch.nn as nn
import torch.optim as optim

# y=2x
X = torch.unsqueeze(torch.linspace(0, 10, 20), 1)
Y = 2 * X

model = nn.Sequential(
    nn.Linear(1, 10),
    nn.ReLU(),
    nn.Linear(10, 1)
)

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = model(X)
    loss = criterion(y_pred, Y)
    loss.backward()
    optimizer.step()

print("Regression ans:")
print(model(torch.tensor([[5.0]])).item())
```
