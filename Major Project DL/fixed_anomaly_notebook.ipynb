## Fix 1: Binary Detector (Sigmoid)

```
import torch
import torch.nn as nn

class BinaryDetector(nn.Module):
    def __init__(self, in_features=128):
        super().__init__()
        self.fc = nn.Linear(in_features, 1)

    def forward(self, x):
        return self.fc(x)  # NO softmax

model = BinaryDetector()
criterion = nn.BCEWithLogitsLoss()

def predict(logits):
    return torch.sigmoid(logits)
```

## Fix 3: False Negative Analysis

```
def get_false_negatives(y_true, y_pred, images):
    fn = []
    for i in range(len(y_true)):
        if y_true[i] == 1 and y_pred[i] == 0:
            fn.append(images[i])
    return fn
```

## Fix 5: Self-Supervised Transformation Model

```
import torch.nn.functional as F

class RotationPredictor(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc = nn.Linear(128, 4)

    def forward(self, x):
        return self.fc(x)

def entropy_score(logits):
    probs = F.softmax(logits, dim=1)
    return -torch.sum(probs * torch.log(probs + 1e-8), dim=1)
```
